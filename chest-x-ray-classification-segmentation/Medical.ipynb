```python
import os
import subprocess
```

```python

Repo_name = "machine-learning-projects"
Repo_url = "https://github.com/Ebenezeragyemang10/machine-learning-projects"
Repo_path = f"/content/{Repo_name}"
Working_subdir = ""
```

```python
if not os.path.isdir(Repo_path):
  "cloning into the repo"
  subprocess.run(["git", "clone", Repo_url])
else:
  print("repo already exist")

target_dir = os.path.join(Repo_path, Working_subdir) if Working_subdir else Repo_path
os.chdir(target_dir)
```
